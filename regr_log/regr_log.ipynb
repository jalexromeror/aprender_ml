# Regresión Logística con SKLearn

### Identificar Sistema Operativo de los usuarios

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn import model_selection
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import seaborn as sb
%matplotlib inline
import warnings
```

### Carga de entrada del archivo CSV.

```python
dataframe = pd.read_csv('/home/yimy/Documentos/Libros/aprender_ml/regr_log/usuarios_win_mac_lin.csv')
dataframe.head()
```

```python
dataframe.describe()
```

```python
print(dataframe.groupby('clase').size())
```

### Visualización de los datos

```python
dataframe[['duracion', 'paginas', 'acciones', 'valor']].hist()
plt.show()
```

```python
sb.pairplot(dataframe.dropna(), hue='clase', height=4, vars=['duracion', 'paginas', 'acciones', 'valor'], kind='reg')
```

### Creación del modelo de regresión logística

```python
X = np.array(dataframe.drop(columns=['clase']))
y = np.array(dataframe['clase'])
X.shape
```

```python
model = linear_model.LogisticRegression(max_iter=1000)
model.fit(X, y)
```

```python
predictions = model.predict(X)
print(predictions[0:5])
```

```python
model.score(X, y)
```

### Validación del modelo

```python
validation_size = 0.20
seed = 7
X_train, X_validation, Y_train, Y_validation = model_selection.train_test_split(X, y, \
    test_size=validation_size, random_state=seed)
```

```python
name = 'Logistic Regression'
kfold = model_selection.KFold(n_splits=10, random_state=seed, shuffle=True)
cv_results = model_selection.cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy')
msg = "%s: %f (%f)" % (name, cv_results.mean(), cv_results.std())
print(msg)
```

```python
predictions = model.predict(X_validation)
print(accuracy_score(Y_validation, predictions))
```

### Reporte de resultados del modelo

```python
print(confusion_matrix(Y_validation, predictions))
```

```python
print(classification_report(Y_validation, predictions))
```
